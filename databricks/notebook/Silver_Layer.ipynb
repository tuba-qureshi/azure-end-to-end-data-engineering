###DATA LOADING

```python
df_procat = spark.read.format('csv')\
    .option("header",True)\
    .option("inferSchema",True)\
    .load('abfss://bronze@storage3datalake.dfs.core.windows.net/AdventureWorks_Product_Categories')
```

```python
df_pro = spark.read.format('csv')\
    .option("header",True)\
    .option("inferSchema",True)\
    .load('abfss://bronze@storage3datalake.dfs.core.windows.net/AdventureWorks_Products')
```

```python
df_ret = spark.read.format('csv')\
    .option("header",True)\
    .option("inferSchema",True)\
    .load('abfss://bronze@storage3datalake.dfs.core.windows.net/AdventureWorks_Returns')
```

```python
df_sales = spark.read.format('csv')\
    .option("header",True)\
    .option("inferSchema",True)\
    .load('abfss://bronze@storage3datalake.dfs.core.windows.net/AdventureWorks_Sales*')
```

```python
df_ter = spark.read.format('csv')\
    .option("header",True)\
    .option("inferSchema",True)\
    .load('abfss://bronze@storage3datalake.dfs.core.windows.net/AdventureWorks_Territories')
```

```python
df_subcat = spark.read.format('csv')\
    .option("header",True)\
    .option("inferSchema",True)\
    .load('abfss://bronze@storage3datalake.dfs.core.windows.net/AdventureWorks_Subcategories')
```

###TRANSFORMATION

###Calendar

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_cal.display()
```

```python
df_cal = df_cal.withColumn('Month',month(col('Date'))) \
    .withColumn('Year',year(col('Date')))
```

```python
df_cal.write.format('parquet') \
            .mode('append') \
            .option("path","abfss://silver@storage3datalake.dfs.core.windows.net/AdventureWorks_Calendar") \
            .save()
```

###Customers

```python
df_cus.display()
```

```python
df_cus = df_cus.withColumn('FullName',concat_ws(' ',col('Prefix'),col('FirstName'),col('LastName')))
```

```python
df_cus.write.format('parquet') \
            .mode('append') \
            .option("path","abfss://silver@storage3datalake.dfs.core.windows.net/AdventureWorks_Customers") \
            .save()
```

###Product Categories

```python
df_procat.write.format('parquet') \
            .mode('append') \
            .option("path","abfss://silver@storage3datalake.dfs.core.windows.net/AdventureWorks_Product_Categories") \
            .save()
```

###Products

```python
df_pro = df_pro.withColumn('ProductSKU',split('ProductSKU','-')[0]) \
    .withColumn('ProductName',split('ProductName',' ')[0])
```

```python
df_pro.write.format('parquet') \
            .mode('append') \
            .option("path","abfss://silver@storage3datalake.dfs.core.windows.net/AdventureWorks_Products") \
            .save()
```

###Return

```python
df_ret.write.format('parquet') \
            .mode('append') \
            .option("path","abfss://silver@storage3datalake.dfs.core.windows.net/AdventureWorks_Returns") \
            .save()
```

### Territories

```python
df_ter.write.format('parquet') \
            .mode('append') \
            .option("path","abfss://silver@storage3datalake.dfs.core.windows.net/AdventureWorks_Territories") \
            .save()
```

###Sales

```python
df_sales = df_sales.withColumn('StockDate',to_timestamp('StockDate'))
```

```python
df_sales = df_sales.withColumn('OrderNumber', regexp_replace(col('OrderNumber'), 'S', 'T'))
```

```python
df_sales = df_sales.withColumn('multiply',col('OrderLineItem')*col('OrderQuantity'))
```

###Sales Analysis

```python
df_sales.groupBy('OrderDate').agg(count('OrderNumber').alias('Total_order')).display()
```
